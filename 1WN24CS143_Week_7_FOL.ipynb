```python

def unify(x, y, subst):
    # Step 1: If both expressions are identical
    if x == y:
        return subst

    # Step 2: If x is a variable
    if isinstance(x, str) and x.islower():
        subst[x] = y
        return subst

    # Step 3: If y is a variable
    if isinstance(y, str) and y.islower():
        subst[y] = x
        return subst

    # Step 4: Check predicate symbols
    if isinstance(x, tuple) and isinstance(y, tuple):

        if x[0] != y[0]:
            return "FAILURE"

        # Step 5: Check number of arguments
        if len(x) != len(y):
            return "FAILURE"

        # Step 6: Unify arguments one by one
        for i in range(1, len(x)):
            result = unify(x[i], y[i], subst)

            if result == "FAILURE":
                return "FAILURE"

        return subst

    return "FAILURE"


# Main program
exp1 = ("Eats", "x", "Apple")
exp2 = ("Eats", "Riya", "y")

SUBST = {}

result = unify(exp1, exp2, SUBST)

if result == "FAILURE":
    print("Unification Failed")
else:
    print("Unification Successful")
    print("Substitution:", result)




```
